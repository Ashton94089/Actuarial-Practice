So far, we have programmed procedurally; now we want to program in an object-oriented way. An object is an instance of a class. A class is a template or blueprint for objects. It defines which attributes (data) and methods (functions) an object will have. Once we have defined a class, we can create any number of objects of that class.

You can think of classes a bit like Plato's ideas. In the class, we describe the properties (attributes) that all objects of that class have and define what can be done with the objects (via the methods).

A simple real-world example would be a class called Car:

But let's look at an example from the insurance industry:

```python
#把「保單資料」跟「能對保單發生的動作」捆綁在一起：
class Policy:

    def __init__(self, policy_number, sum_insured):
        # 這些是屬性（Attribute），每個建立出來的保單都會保存自己的資料
        self.policy_number = policy_number
        self.sum_insured = sum_insured
        self.is_active = True

    # 這些是方法（Method），其實就是寫在 Class 裡面的 Function
    def calculate_premium(self, rate):
        return self.sum_insured * rate

    def cancel(self):
        self.is_active = False
        print(f"保單 {self.policy_number} 已失效")


# 使用時：
my_policy = Policy("POL12345", 1000000)  # 建立一個實體物件
print(my_policy.calculate_premium(0.002))  # 直接對這個物件操作
my_policy.cancel()  # 改變物件內部的狀態 (is_active 變成 False)
```

```python
class InsuranceContract:

    def __init__(
        self, contract_number, insured_sum, holder
    ):  # So-called constructor with attributes
        self.contract_number = contract_number
        self.insured_sum = insured_sum
        self.holder = holder

    def calculate_premium(self):
        # Here we use a simple rule to calculate the premium
        return self.insured_sum * 0.05
```

```python
# Creating an insurance contract
contract1 = InsuranceContract("12345", 100000, "Max Mustermann")
contract2 = InsuranceContract("12346", 200000, "Petra Musterfrau")

# Accessing the attributes of the object
print(f"Contract number: {contract1.contract_number}")
print(f"Insured sum: {contract1.insured_sum}")
print(f"Policyholder: {contract1.holder}")

# Calculating the premium
premium1 = contract1.calculate_premium()
print(f"The calculated premium is: {premium1:.2f} Euros")

premium2 = contract2.calculate_premium()
print(f"The calculated premium is: {premium2:.2f} Euros")
```

Introduction to Inheritance (繼承)

With object-oriented programming, we can already modularize and reuse large portions of our code by working with classes and objects. By encapsulating properties and behaviors within a class, we can avoid redundant code fragments and improve maintainability.

But object-oriented programming offers even more: Inheritance is a central concept that allows us to link classes and establish relationships between them. Inheritance allows us to define a general class from which specialized classes can be derived. This not only reduces the effort but also creates a clearer structure.

**Advantages of inheritance:**

* Code reuse: Common functionalities only need to be defined once in a base class. Derived classes automatically inherit this functionality.

* Extensibility: New, specialized classes can be easily created by inheriting from an existing class and adding additional functions or attributes as needed.

* Maintainability: A clearly structured inheritance hierarchy makes the code more readable and easier to maintain.

Example: Inheritance for Policyholders

Imagine we want to manage other types of people in our system besides policyholders, such as brokers or agents. They all share common attributes (e.g., name, date of birth, address), but also have specific differences. Instead of redefining these properties and methods in each class, we can create a general class, `Person`, from which all the more specialized classes inherit.

Let's try this out:

```python
# #在 Python 中，def __init__(...) -> None: 裡面的 -> None 叫做型別提示（Type Hinting / Return Type Annotation）。
# #在 Python 中，類別的建構子（__init__ 方法）的作用是初始化物件，它在語法上規定不能有任何 return 回傳值（或者只能寫 return None）
# #也可以不寫

# # 以下兩種寫法在 Python 執行的結果完全一模一樣：

# # 寫法 A：帶有型別提示
# def __init__(self, pName, pDateOfBirth, pAddress) -> None:
#     self.name = pName

# # 寫法 B：不帶型別提示
# def __init__(self, pName, pDateOfBirth, pAddress):
#     self.name = pName
```

```python
# #除了 None 之外，其他函式也可以用 -> 來標示回傳的資料型態：
# def add_numbers(a: int, b: int) -> int:
#     return a + b  # 標示這個函式接受兩個整數 (int)，並會回傳一個整數 (int)

# def get_name(self) -> str:
#     return self.name  # 標示這個函式會回傳字串 (str)
```

```python
class Person():
  def __init__(self, pName, pDateOfBirth, pAddress) -> None:
    self.name = pName
    self.date_of_birth = pDateOfBirth
    self.address = pAddress
```

Now we want the policyholder class to inherit from the person class:

```python
class Policyholder(Person): # that's it, just wrap the class in parentheses, boom, inherited.
  def __init__(self, pName, pDateOfBirth, pAddress, pPolicyNumber) -> None:
    super().__init__(pName, pDateOfBirth, pAddress)
    self.policy_number = pPolicyNumber

  def display(self):
    print(f"Policyholder: {self.name}")
    print(f"Date of Birth: {self.date_of_birth}")
    print(f"Address: {self.address}")
    print(f"Policy Number: {self.policy_number}")
```

```python
# let's try it out:
customer = Policyholder("Max Mustermann", "01.01.1980", "Musterstraße 1", "123456789")
customer.display()
```

# **Special Class Methods**

We have already seen that there are special methods on classes in Python. In principle, every class should implement the following standard methods:

`__init__()` -> Constructor

`__repr__()` -> a (technical) string representation of the object

`__str__()` -> a (human-readable) string representation of the object

`__eq__()` -> allows two objects of the class to be checked for equality using "=="

`__hash__()` -> returns a unique hash value for each instance

Further standard methods (which do not make sense for Person objects) are:

`__add__()` -> add/concatenate two objects using "+"

`__sub__()` -> subtract two objects using "-"  # Note: The standard Python method is __sub__, not `__subtract__`

`__getitem__(idx)` -> index access returns the value at idx

`__setitem__(idx, value)` -> sets the element at index idx to the value

`__len__()` -> returns the length of the object

`__del__()` -> deletes the instance of the object (destructor)

`__iter__()` -> for iterating over the object

We are implementing the first ones:

```python
class Person:
  def __init__(self, name, date_of_birth, address):
    self.name = name
    self.date_of_birth = date_of_birth
    self.address = address

  def display_details(self):
    print(f"Name: {self.name}")
    print(f"Date of Birth: {self.date_of_birth}")
    print(f"Address: {self.address}")

  def __str__(self) -> str:
    # Returns a user-friendly description of the person
    return f"Person(Name: {self.name}, Date of Birth: {self.date_of_birth}, Address: {self.address})"

  def __repr__(self) -> str:
    # Returns a detailed, "technical" description useful for developers
    return f"Person(name={self.name!r}, date_of_birth={self.date_of_birth!r}, address={self.address!r})"

  def __eq__(self, other: object) -> bool:
    # Checks if two Person objects are equal (same name, date of birth, and address)
    if isinstance(other, Person):
      return (self.name == other.name and
           self.date_of_birth == other.date_of_birth and
           self.address == other.address)
    return False

  def __hash__(self) -> int:
    # Generates a hash value based on name, date of birth, and address
    return hash((self.name, self.date_of_birth, self.address))
```

Circle

Create a subclass `Circle` that inherits from `Shape`. Implement a constructor that sets the shape's name to "Circle" and stores its radius. Override the `area` method to calculate the circle's area. Also implement a `__eq__()` method to compare two circles.

Note: `math.pi` returns the value of the radius π.

```python
import math

class Circle(Shape):
  def __init__(self, radius):
    super().__init__("Circle") #Sets the name of the shape to "Circle"
    self.radius = radius

  def area(self):
    #Calculates the area of the circle
    return math.pi*(self.radius**2)

  def __eq__(self, otherCircle):
    #Compares whether two circles are equal (same radius)
    if isinstance(otherCircle, Circle):
      return self.radius == otherCircle.radius
    return NotImplemented

  def __repr__(self):
    #Returns a detailed, technical description
    return f"Circle(radius={self.radius!r})"

  def __str__(self):
    #Returns a user-friendly description of the circle
    return f"Circle with radius: {self.radius}"
```

```python
#Example:
circle = Circle(5)
print(circle.name)  #Output: Circle
print(circle.area()) #Output: 78.53981633974483

circle2 = Circle(6)
circle3 = Circle(5)

print(circle == circle2) #False
print(circle == circle3) #True
```

Rectangle

Create a subclass `Rectangle` that inherits from `Shape`. Implement the constructor that sets the name and stores the width and height. Override the `area` method to calculate the area of ​​the rectangle.

```python
class Rectangle(Shape):
  def __init__(self, width, height):
    super().__init__("Rectangle") #Sets the name of the shape to "Rectangle"
    self.width = width
    self.height = height

  def area(self):
    #Calculates the area of the rectangle
    return self.width*self.height

  def __repr__(self):
    #Returns a detailed, technical string representation
    return f"Rectangle(width={self.width!r}, height={self.height!r})"

  def __str__(self):
    # Returns a user-friendly string representation of the rectangle
    return f"Rectangle with width: {self.width} and height: {self.height}"

  def __eq__(self, otherRectangle):
    # Checks if two rectangle are equal (same width and height)
    if isinstance(otherRectangle, Rectangle):
      return self.width == otherRectangle.width and self.height == otherRectangle.height
    return NotImplemented
```

```python
retangle = Rectangle(3,5)
print(retangle.name)
print(retangle.area())
print(repr(retangle))
print(str(retangle))
```

Square

Create a subclass `Square` that inherits from `Rectangle`. Implement a constructor to store the side length and name, and override the `area` method to calculate the square's area. Also, create a `__eq__()` method to test two squares for equality.

```python
class Square(Rectangle):
  def __init__(self, side_length):
    super().__init__(side_length, side_length) # Sets the width and height equal to the side length
    self.name = "Square" # Sets the name of the shape to "Square"

  def area(self):
    #Calculates the area of the square
    return self.width*self.height

  def __eq__(self, otherSquare):
    #Checks if two squares are equal (same side length)
    if isinstance(otherSquare, Square):
      return self.width == otherSquare.width
    return NotImplemented

  def __str__(self):
    return f"Square with side length: {self.width}"
```

```python
square = Square(4)
print(square.name)
print(square.area())

print(square == Square(3))
print(square == Square(4))
```
