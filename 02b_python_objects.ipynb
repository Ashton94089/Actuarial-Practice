Built-in Data Types in Python

In addition to user-defined classes, Python offers a variety of built-in data types that are extremely useful in everyday programming. These data types are also based on classes, meaning they have specific methods and properties that we can work with directly.

The most important built-in data types include:

* Strings: A string that stores text.

* Lists: An ordered collection of items.

* Sets: An unordered collection of unique items.

* Dictionaries: A collection of key-value pairs.

These data types are predefined in Python and can be used directly without having to create them yourself. They are very flexible and allow us to solve many tasks efficiently.

In the following, we will look at how these data types work and how we can use them in our programs.

Strings belong to the sequence type; https://docs.python.org/3.5/library/stdtypes.html#typesseq

Also included are ranges, tuples (both immutable), and lists (mutable).

All immutable sequences can perform the [common sequence operations](https://docs.python.org/3/library/stdtypes.html#common-sequence-operations), and mutable sequences can additionally perform the [mutable sequence operations](https://docs.python.org/3/library/stdtypes.html#mutable-sequence-types).

```python
#indexing
greeting = 'hello there'
(
    greeting[0],
    greeting[5],
    len(greeting),
    greeting[len(greeting) - 1]
)
```

```python
#negative indexes
(
    greeting[-1],
    greeting[-2],
    greeting[-len(greeting)]
)
```

```python
# "slices"
(
    greeting[0:11],
    greeting[0:5],
    greeting[6:11]
)
```

```python
# default slice ranges
(
    greeting[:11],
    greeting[6:],
    greeting[:]
)
```

```python
# slice "steps"
(
    greeting[0:11:2],
    greeting[::3],
    greeting[6:11:2]
)
```

```python
# negative steps
greeting[::-1]
```

```python
# other sequence ops
greeting = 'hello there!'
(
    greeting.count('e'),    # 統計數量：計算 'e' 出現的總次數 -> 結果為 3
    greeting.index('e'),    # 尋找索引：找出第一個 'e' 出現的位置 index -> 結果為 1
    greeting.index('e', 2),  # 指定起點搜尋：從 index 2 開始往後找第一個 'e' 的位置 -> 結果為 8
    'e' in greeting,     # 成員檢查：檢查 'e' 是否在字串中 -> 結果為 True
    'ü' not in greeting,  # 反向成員檢查：檢查 'ü' 是否不在字串中 -> 結果為 True
    min(greeting),       # 最小值：依 Unicode 碼挑出最小字元（空白鍵 ' ' 的 Unicode 最小）-> 結果為 ' '
    max(greeting)       # 最大值：依 Unicode 码挑出最大字元（字母 't' 的 Unicode 最大）-> 結果為 't'
)
```

```python
adjective = 'frigid'
adverb = 'hastily'
noun = 'Alfred'
number = 8
verb = 'eat'

sentence = f'It was a {adjective} day when {noun} decided to {adverb} {verb} {number*100} lines of code.'

print(sentence)
```

Task 1 (Palindrome Check) (迴文檢查)

Write a program that checks whether a given string is a palindrome. A palindrome is a word or sentence that reads the same forwards and backwards (ignoring spaces and case sensitivity).

Examples of palindromes:

* Eine Horde bedrohe nie. (德文：永遠不要威脅一群人)

* A man a plan a canal panama.

* EINE TREUE FAMILIE BEI LIMA FEUERTE NIE

```python
# Initialization
text = "Eine Horde bedrohe nie"

# Palindrome Check
# 1. Convert all characters to lowercase
# 2. Remove spaces
# 3. Compare the processed string with its reverse

# Normalize the text: convert to lowercase and remove spaces
normalized_text = ''.join(text.lower().split())

# Check whether the normalized text is a palindrome
ist_palindrom = normalized_text == normalized_text[::-1]

print(f"'{text}' is a palindrome: {ist_palindrom}")
```

```python
# Run this code once to fully load the book:
import urllib.request

hegel_url = 'https://www.gutenberg.org/cache/epub/6698/pg6698.txt'
pdg_text = urllib.request.urlopen(hegel_url).read().decode()

# urllib.request.urlopen(hegel_url)：
# 向指定的 URL 發送 HTTP 請求，開啟網路連接以下載檔案（這裡下載的是古騰堡計劃中黑格爾的《精神現象學》文本）。

# .read()：讀取下載下來的原始二進位資料（bytes）。

# .decode()：
# 將二進位資料解碼轉換為 UTF-8 格式的文字字串，存入 pdg_text 變數中。
```

How do you get from the individual characters of pdg_text to the words of the Phenomenology of Spirit? (Hint: the split function might help.)

```python
pdg_words = pdg_text.lower().split()
len(pdg_words)
```

And now to the unambiguous words?

```python
pdg_words_unique = set(pdg_words)
len(pdg_words_unique)
```

And now for the frequency of individual words. How often was each word written? (Hint: use a dictionary.)

```python
pdg_word_count = {}

for word in pdg_words:
    if word in pdg_word_count:
        pdg_word_count[word] += 1
    else:
        pdg_word_count[word] = 1
```
