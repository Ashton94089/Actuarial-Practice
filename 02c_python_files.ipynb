File Operations in Python

In Python, we can easily work with files, whether to read data or write information. Working with files involves three steps:

1. Opening the file: We can open a file using the `open()` function to read or write to it.

2. Reading or writing: Once the file is open, we can perform the desired operations.

3. Closing the file: After the operations are complete, the file is closed to release resources.

Basic file operations:

* Reading a file: We can read the contents of a file using `read()` or `readline()`.

* Writing a file: We can write text to a file using `write()`.

* Working with file paths: Absolute and relative paths can be used to save or open files in different locations.

2. Writing to a file

To write text to a file, we open the file in write mode 'w'.

```python
# 建立並寫入 example.txt
data = "Hello Google Colab!\n這是一份測試用的文字檔案。"
with open('example.txt', 'w', encoding='utf-8') as file:
  file.write(data)
```

Code Examples:

1. Reading a File

This example reads the entire contents of a text file.

```python
# Open a file in read mode ('r)
with open('example.txt','r') as file:
  contents = file.read()
  print(contents)
```

我們改用 'a'（Append，附加模式）往檔案後面寫入新內容，這樣就不會蓋掉原本寫好的文字：

```python
# 在原本的 example.txt 後面追加更多內容
with open('example.txt', 'a', encoding='utf-8') as file:
    file.write("\n這是第二行：Python 逐行讀取測試。")
    file.write("\n這是第三行：黑格爾的精神現象學。")
    file.write("\n這是第四行：End of file.")
```

3. Reading a file line by line

Sometimes it is more efficient to read the file line by line.

```python
# Open the file and read it line by line
# Open the file for reading
with open('example.txt', 'r') as file:
    # Read each line and print it
    for line in file:
        print(line, end='')
        # print(line.strip())
```

4. Working with file paths

We can also write and read files in locations other than the current folder:

```python
import os

# 自動檢查並建立 images 資料夾
os.makedirs('./images', exist_ok=True)

data = "This is sample text for another file.\nHere is a new line."

with open("./images/output.txt", "w") as file:
  file.write(data)
```

```python
with open("./images/output.txt", "r") as file:
  for line in file:
    print(line.strip())
```
