Introduction to Pandas

Pandas is one of the most popular libraries for data analysis and manipulation in Python. It was specifically designed to work efficiently and easily with structured data, similar to spreadsheet applications like Excel.

Pandas offers two main structures for handling data:

* Series: A one-dimensional data structure, similar to a list or an array.

* DataFrame: A two-dimensional data structure, comparable to a table in a database or an Excel spreadsheet.

Why Pandas?

Pandas is especially useful when it comes to ingesting, processing, and analyzing large amounts of data. Typical tasks that Pandas simplifies include:

* Ingesting data from various formats (e.g., CSV, Excel, SQL databases).
* Efficiently inspecting and analyzing data.

* Data cleansing and manipulation (e.g., handling missing values, renaming columns).

* Transforming and aggregating data for further analysis or visualization.

Thanks to its broad functionality and ease of use, Pandas has become a standard tool for data analysis, particularly in data science, financial analysis, and insurance.

Below, we will look at the key features of Pandas and learn how to efficiently ingest, analyze, and clean data.

Criticisms of Pandas

Although Pandas is a powerful library, there are some criticisms to keep in mind:

1. Memory and computational overhead: Pandas is memory-intensive because it loads data into RAM. With very large datasets, this can lead to performance issues. For large datasets that don't fit in RAM, alternative tools like Dask, PySpark, or Apache Arrow, which are optimized for distributed processing, should be considered.

2. Complexity with large data pipelines: Pandas is very efficient for smaller projects and analyses. However, with larger data pipelines or complex data analyses, the code can become unwieldy and difficult to maintain. In particular, dealing with many nested Pandas functions can make the code difficult to understand.

Despite these limitations, Pandas remains a powerful tool for most data analyses, especially with medium-sized datasets.

Another example is (synthetic) stock price charts:

````python
# Parameters for the simulation
# 固定隨機亂數種子，確保每次執行產生的隨機數都一樣（方便重複實驗）
np.random.seed(42)
# 生成從 2023-01-01 開始，連續 365 天的時間序列 index
days = pd.date_range(start='2023-01-01', periods=365, freq='D')
# 股票初始價格 $S_0 = 100$
starting_price = 100 # Starting price of the stock
# 波動率（標準差 $\sigma$）= 2%
volatility = 0.02 # Stock volatility (2%)
# 期望日報酬率（平均值 $\mu$）= 0.1%
returns = 0.001 # Expected daily return (0.1%)

# Generate daily changes based on Brownian motion
# 1. 根據常態分配生成每天的對數報酬率變化量
changes = np.random.normal(loc=returns, scale=volatility, size=len(days))

# Calculate the stock prices using the cumulative sum
# 2. 利用對數報酬率的「累加（cumsum）」與「指數（exp）」計算每天的股價
prices = starting_price * np.exp(np.cumsum(changes))

# Create the Pandas Series
# 3. 將計算好的價格陣列包裝成 Pandas Series，並以日期作為 index
stock_price = pd.Series(prices, index=days)

# Output of the first few days of the stock price chart
# 4. 印出前 20 天的股價結果
print(stock_price.head(20))
````

Here is another example of time-dependent (synthetic) data:

````python
import numpy as np
import pandas as pd

# Parameters for the simulation
np.random.seed(42)
days = pd.date_range(start='2023-01-01', periods=365, freq='D')

# Stock parameters for different companies
companies = {
    'Apple':{'start_price':150,'volatility':0.02,'return':0.001},
    'Nvidia':{'start_price':200,'volatility':0.025,'return':0.0012},
    'Google':{'start_price':180,'volatility':0.018,'return':0.0008},
    'SAP':{'start_price':120,'volatility':0.015,'return':0.0009},
    'Microsoft':{'start_price':250,'volatility':0.02,'return':0.0011},
    'Tesla':{'start_price':200,'volatility':0.03,'return':0.0015},
}

# Create a DataFrame to store all stock price histories
stock_histories = pd.DataFrame(index=days)

# Simulation of stock prices for each company
for company, parameters in companies.items():
  start_price = parameters['start_price']
  volatility = parameters['volatility']
  daily_return = parameters['return']

  # Generate daily changes based on Brownian motion
  changes = np.random.normal(loc=daily_return, scale=volatility, size=len(days))

  # Calculate stock prices using cumulative sum
  prices = start_price * np.exp(np.cumsum(changes))

  # Add the simulation as a series into the DataFrame
  stock_histories[company] = prices

# Output the first 20 days of simulated stock price histories
stock_histories.head(20)
````

# Grouping Data and Aggregation Functions in Pandas

In data analysis, it is often necessary to group data in order to calculate aggregate statistics. Pandas offers powerful functions for grouping data and performing various aggregation operations.

# Fundamental Concepts

* Grouping: Data is grouped using the `groupby()` method. This method splits the data into groups based on the values ​​of one or more columns.
* Aggregation: After grouping, various aggregation functions can be applied to calculate statistical metrics. Commonly used aggregation functions include:

  * `mean()`: Calculates the average.
  * `sum()`: Calculates the sum.
  * `count()`: Counts the number of elements.
  * `max()`: Determines the maximum value.
  * `min()`: Determines the minimum value.

# Example: Aggregating loss data

Suppose we have a DataFrame named `damage_data` containing loss information, with a column for the region and one for the loss amount. We can calculate the average loss per region as follows:

```
# Calculate average damage/claim amount by region
average_damage = damage_data.groupby('Region')['Claim_Amount'].mean()
print(average_damage)
```

7. Distribution of age groups

Divide the passengers into different age groups (e.g., children, adolescents, adults, seniors) and examine how these groups affect the survival rate.

````python
# Definition of age groups
bins = [0, 3, 12, 18, 60, 120] # Age limits for the groups
labels = ['Toddlers', 'Children', 'Adolescents', 'Adults', 'Seniors']  # Labels for the groups

# Creating a new column 'AgeGroup' with the age groups
titanic_df['AgeGroup'] = pd.cut(titanic_df['Age'], bins=bins, labels=labels, right=False)

# Calculating the survival rate for each age group
survival_rate_by_age_group  = titanic_df.groupby('AgeGroup')['Survived'].mean()

# Output of results
# print(f"Survival rate by age group:{survival_rate_by_age_group}")
print("Survival rate by age group:")
print(survival_rate_by_age_group)


# 做法 A：維持現在的行為（推薦）
# 如果你希望所有切分好的年齡層（就算該區間沒人）都顯示出來：
# survival_rate_by_age_group = titanic_df.groupby('AgeGroup', observed=False)['Survived'].mean()

# 做法 B：採用未來的預設行為
# 如果你只想計算與顯示「真的有乘客存在」的年齡層：
# survival_rate_by_age_group = titanic_df.groupby('AgeGroup', observed=True)['Survived'].mean()
````
