```python
print("Example of 'break':")
for i in range(1, 11):
    if i == 5:
        print("The loop is interrupted at i = 5.")
        break  # Loop ends as soon as i is 5
    print(i)
```

```python
print("\nExample of 'continue':")
for i in range(1,11):
  if i%2==0:
    continue # Skips all even numbers
  print(i) #Only odd numbers are displayed
```

```python
print("\nExample of 'pass':")
for i in range(1, 11):
    if i % 2 == 0:
        pass  # Placeholder that does nothing
    else:
        print(i)  # Outputs only odd numbers
```

```python
# Outputs a multiplication table of the basic times tables
# Initialization
n=10
# Calculation and display of the multiplication table
for i in range(1,n+1):
  for j in range(1,n+1):
    z=i*j
    print(f"{z:4}",end="")
  print()
```

```python
# Initialization
n = 30

# Counting and condition checking
for i in range(1,n+1):
    if (i % 15) == 0:
        print("FizzBuzz")
    elif (i % 3) == 0:
        print("Fizz")
        continue
    elif (i % 5) == 0:
        print("Buzz")
    else:
        print(i)
```

```python
n=15
for i in range(2,n+1):
  is_prime=True
  for j in range(2,i):
    if i%j==0:
      is_prime=False
      break
  if is_prime:
    print(i)
```
